# RUL Prediction — NASA IMS Bearing Dataset
### Using PCA + Health Indicator + Threshold + RUL

## Step 1: Load Data

NASA IMS Bearing dataset (2nd_test).
Each file = one time snapshot of all 4 bearings.
We pick **Bearing 1** (column 0).
Sampling frequency = 20,000 Hz.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.stats import skew, kurtosis
from scipy.fft import fft, fftfreq
```

```python
# ── Load NASA IMS bearing dataset ────────────────────────────────────────────
# Option A: kagglehub auto-downloads the dataset if available
# Option B: set data_dir manually to your local 2nd_test folder path
try:
    import kagglehub
    _base = kagglehub.dataset_download("vinayak123tyagi/bearing-dataset")
    from pathlib import Path as _P
    data_dir = _P(_base) / "2nd_test" / "2nd_test"
    if not data_dir.exists():
        data_dir = _P(_base) / "2nd_test"
except Exception:
    from pathlib import Path
    data_dir = Path("./2nd_test")   # ← set your local path here if needed

Fs = 20000
bearing_col = 0
data_paths = sorted(data_dir.glob("*"))
print("Total files:", len(data_paths))
print("First file :", data_paths[0].name)
print("Last file  :", data_paths[-1].name)
```

```python

all_segments = []
for fp in data_paths:
    raw = np.loadtxt(fp)
    all_segments.append(raw[:, bearing_col])

vibration = np.concatenate(all_segments)
time = np.arange(len(vibration)) / Fs

print("Vibration shape:", vibration.shape)
print("Total duration :", round(time[-1], 1), "seconds")
```

```python
plt.figure(figsize=(10,4))
plt.plot(time, vibration)
plt.title("Raw Vibration Signal — Bearing 1 (full run)")
plt.xlabel("Time (seconds)")
plt.ylabel("Acceleration")
plt.grid(True)
plt.show()
```

```python
print("Sampling Frequency:", Fs, "Hz")
print("Samples per file  :", len(all_segments[0]))
print("Number of files   :", len(data_paths))
```

## Step 2: Windowing

We divide the long vibration signal into windows.
Each window = one sample. From each window we extract features.

One file = ~1 second of data.
We use **window = 1 file size** so each window = one snapshot.

```python
fs = 20000
window_seconds = 5
window_size = fs * window_seconds  
step = window_size // 2

print("Window size (samples):", window_size)
print("Window size (seconds):", window_seconds)
print("Step size (samples):", step)
```

```python
time_final = []

for i in range(0, len(vibration) - window_size, step):
    t = time[i + window_size // 2]
    time_final.append(t)

time_final = np.array(time_final)

print("Total windows:", len(time_final))
```

## Step 3: Feature Extraction

We extract 3 types of features from each window:
- **Time Domain** — 12 features
- **Frequency Domain** — 10 features
- **Time-Frequency (Wavelet)** — 5 features

Total = 27 features per window

### 3A — Time Domain Features (12)

```python
time_features = {}
```

```python
# 1. RMS
rms = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    rms.append(np.sqrt(np.mean(seg**2)))
rms = np.array(rms)
time_features["RMS"] = rms

plt.figure(figsize=(10,4))
plt.plot(time_final, rms)
plt.title("RMS — Energy of vibration")
plt.xlabel("Time (s)"); plt.ylabel("RMS")
plt.grid(True)
plt.show()
```

```python
# 2. Mean
mean_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    mean_vals.append(np.mean(seg))
mean_vals = np.array(mean_vals)
time_features["Mean"] = mean_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, mean_vals)
plt.title("Mean — Average vibration")
plt.xlabel("Time (s)"); plt.ylabel("Mean")
plt.grid(True)
plt.show()
```

```python
# 3. Standard Deviation
std_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    std_vals.append(np.std(seg))
std_vals = np.array(std_vals)
time_features["Std"] = std_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, std_vals)
plt.title("Standard Deviation — Spread of vibration")
plt.xlabel("Time (s)"); plt.ylabel("Std")
plt.grid(True)
plt.show()
```

```python
# 4. Variance
var_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    var_vals.append(np.var(seg))
var_vals = np.array(var_vals)
time_features["Variance"] = var_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, var_vals)
plt.title("Variance — Instability of vibration")
plt.xlabel("Time (s)"); plt.ylabel("Variance")
plt.grid(True)
plt.show()
```

```python
# 5. Skewness
skew_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    skew_vals.append(skew(seg))
skew_vals = np.array(skew_vals)
time_features["Skewness"] = skew_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, skew_vals)
plt.title("Skewness — Asymmetry of vibration")
plt.xlabel("Time (s)"); plt.ylabel("Skewness")
plt.grid(True)
plt.show()
```

```python
# 6. Kurtosis
kurt_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    kurt_vals.append(kurtosis(seg))
kurt_vals = np.array(kurt_vals)
time_features["Kurtosis"] = kurt_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, kurt_vals)
plt.title("Kurtosis — Sharpness / impulsiveness")
plt.xlabel("Time (s)"); plt.ylabel("Kurtosis")
plt.grid(True)
plt.show()
```

```python
# 7. Peak
peak_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    peak_vals.append(np.max(np.abs(seg)))
peak_vals = np.array(peak_vals)
time_features["Peak"] = peak_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, peak_vals)
plt.title("Peak — Maximum vibration amplitude")
plt.xlabel("Time (s)"); plt.ylabel("Peak")
plt.grid(True)
plt.show()
```

```python
# 8. Peak to Peak
ptp_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    ptp_vals.append(np.max(seg) - np.min(seg))
ptp_vals = np.array(ptp_vals)
time_features["Peak2Peak"] = ptp_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, ptp_vals)
plt.title("Peak to Peak — Total swing of vibration")
plt.xlabel("Time (s)"); plt.ylabel("Peak2Peak")
plt.grid(True)
plt.show()
```

```python
# 9. Crest Factor
crest_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    rms_seg = np.sqrt(np.mean(seg**2))
    crest_vals.append(np.max(np.abs(seg)) / (rms_seg + 1e-12))
crest_vals = np.array(crest_vals)
time_features["CrestFactor"] = crest_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, crest_vals)
plt.title("Crest Factor — Peak / RMS")
plt.xlabel("Time (s)"); plt.ylabel("Crest Factor")
plt.grid(True)
plt.show()
```

```python
# 10. Mean Absolute Value (MAV)
mav_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    mav_vals.append(np.mean(np.abs(seg)))
mav_vals = np.array(mav_vals)
time_features["MAV"] = mav_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, mav_vals)
plt.title("MAV — Mean Absolute Value")
plt.xlabel("Time (s)"); plt.ylabel("MAV")
plt.grid(True)
plt.show()
```

```python
# 11. Shape Factor (RMS / MAV)
sf_vals = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    rms_seg = np.sqrt(np.mean(seg**2))
    mav_seg = np.mean(np.abs(seg))
    sf_vals.append(rms_seg / (mav_seg + 1e-12))
sf_vals = np.array(sf_vals)
time_features["ShapeFactor"] = sf_vals

plt.figure(figsize=(10,4))
plt.plot(time_final, sf_vals)
plt.title("Shape Factor — RMS / MAV")
plt.xlabel("Time (s)"); plt.ylabel("Shape Factor")
plt.grid(True)
plt.show()
```

```python

cf=[]
count=0

for i in range(0,len(vibration)-window_size,step):
    segment=vibration[i:i+window_size]
    cf_val=np.max(np.abs(segment))/(np.mean(np.sqrt(np.abs(segment)))**2)
    cf.append(cf_val)
cf=np.array(cf)
time_features["ClearanceFactor"] =cf


plt.figure(figsize=(10,5))
plt.plot(time_final,cf)
plt.xlabel("Time (seconds)")
plt.ylabel("Clearance Factor")
plt.title("Bearing looseness of vibration(10s window, 50% overlap)")
plt.grid(True)
plt.show()
```

```python
# Build time-domain DataFrame
time_df = pd.DataFrame(time_features)
time_df["time"] = time_final
print("Time domain features shape:", time_df.shape)
time_df.head()
```

### 3B — Frequency Domain Features (10)

```python
freq_features = {}
```

```python
# 1. Spectral Entropy
spec_entropy = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    psd = spectrum**2
    psd_norm = psd / (np.sum(psd) + 1e-12)
    entropy = -np.sum(psd_norm * np.log2(psd_norm + 1e-12))
    spec_entropy.append(entropy)
spec_entropy = np.array(spec_entropy)
freq_features["SpectralEntropy"] = spec_entropy

plt.figure(figsize=(10,4))
plt.plot(time_final, spec_entropy)
plt.title("Spectral Entropy — How spread is the energy")
plt.xlabel("Time (s)"); plt.ylabel("Spectral Entropy")
plt.grid(True)
plt.show()
```

```python
# 2. Spectral Kurtosis
spec_kurt = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    spec_kurt.append(kurtosis(spectrum))
spec_kurt = np.array(spec_kurt)
freq_features["SpectralKurtosis"] = spec_kurt

plt.figure(figsize=(10,4))
plt.plot(time_final, spec_kurt)
plt.title("Spectral Kurtosis — Spikiness in frequency domain")
plt.xlabel("Time (s)"); plt.ylabel("Spectral Kurtosis")
plt.grid(True)
plt.show()
```

```python
# 3. Spectral Skewness
spec_skew = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    spec_skew.append(skew(spectrum))
spec_skew = np.array(spec_skew)
freq_features["SpectralSkewness"] = spec_skew

plt.figure(figsize=(10,4))
plt.plot(time_final, spec_skew)
plt.title("Spectral Skewness")
plt.xlabel("Time (s)"); plt.ylabel("Spectral Skewness")
plt.grid(True)
plt.show()
```

```python
# 4. Peak Frequency
peak_freq = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    freqs    = fftfreq(window_size, d=1/Fs)[:window_size//2]
    peak_freq.append(freqs[np.argmax(spectrum)])
peak_freq = np.array(peak_freq)
freq_features["PeakFreq"] = peak_freq

plt.figure(figsize=(10,4))
plt.plot(time_final, peak_freq)
plt.title("Peak Frequency — Dominant frequency")
plt.xlabel("Time (s)"); plt.ylabel("Frequency (Hz)")
plt.grid(True)
plt.show()
```

```python
# 5. Spectral RMS
spec_rms = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    spec_rms.append(np.sqrt(np.mean(spectrum**2)))
spec_rms = np.array(spec_rms)
freq_features["SpectralRMS"] = spec_rms

plt.figure(figsize=(10,4))
plt.plot(time_final, spec_rms)
plt.title("Spectral RMS — Overall spectral magnitude")
plt.xlabel("Time (s)"); plt.ylabel("Spectral RMS")
plt.grid(True)
plt.show()
```

```python
# 6. Spectral Centroid
spec_centroid = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    freqs    = fftfreq(window_size, d=1/Fs)[:window_size//2]
    psd      = spectrum**2
    centroid = np.sum(freqs * psd) / (np.sum(psd) + 1e-12)
    spec_centroid.append(centroid)
spec_centroid = np.array(spec_centroid)
freq_features["SpectralCentroid"] = spec_centroid

plt.figure(figsize=(10,4))
plt.plot(time_final, spec_centroid)
plt.title("Spectral Centroid — Weighted mean frequency")
plt.xlabel("Time (s)"); plt.ylabel("Centroid (Hz)")
plt.grid(True)
plt.show()
```

```python
# 7. Spectral Energy
spec_energy = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    spec_energy.append(np.sum(spectrum**2))
spec_energy = np.array(spec_energy)
freq_features["SpectralEnergy"] = spec_energy

plt.figure(figsize=(10,4))
plt.plot(time_final, spec_energy)
plt.title("Spectral Energy")
plt.xlabel("Time (s)"); plt.ylabel("Spectral Energy")
plt.grid(True)
plt.show()
```

```python
# 8. Frequency Variance
freq_var = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    freqs    = fftfreq(window_size, d=1/Fs)[:window_size//2]
    psd      = spectrum**2
    centroid = np.sum(freqs * psd) / (np.sum(psd) + 1e-12)
    f_var    = np.sum(((freqs - centroid)**2) * psd) / (np.sum(psd) + 1e-12)
    freq_var.append(f_var)
freq_var = np.array(freq_var)
freq_features["FreqVariance"] = freq_var

plt.figure(figsize=(10,4))
plt.plot(time_final, freq_var)
plt.title("Frequency Variance — Spread of spectral energy")
plt.xlabel("Time (s)"); plt.ylabel("Freq Variance")
plt.grid(True)
plt.show()
```

```python
# 9. Band Power 1X (1 × shaft speed = 30 Hz band ±5 Hz)
shaft_hz = 30  

bp1x = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    freqs    = fftfreq(window_size, d=1/Fs)[:window_size//2]
    psd      = spectrum**2
    mask     = (freqs >= shaft_hz - 5) & (freqs <= shaft_hz + 5)
    bp1x.append(np.sum(psd[mask]) if mask.any() else 0.0)
bp1x = np.array(bp1x)
freq_features["BandPower1X"] = bp1x

plt.figure(figsize=(10,4))
plt.plot(time_final, bp1x)
plt.title("Band Power 1X (1× shaft frequency)")
plt.xlabel("Time (s)"); plt.ylabel("Band Power")
plt.grid(True)
plt.show()
```

```python
# 10. Band Power 2X (2 × shaft speed)
bp2x = []
for i in range(0, len(vibration) - window_size, step):
    seg = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    freqs    = fftfreq(window_size, d=1/Fs)[:window_size//2]
    psd      = spectrum**2
    mask     = (freqs >= 2*shaft_hz - 5) & (freqs <= 2*shaft_hz + 5)
    bp2x.append(np.sum(psd[mask]) if mask.any() else 0.0)
bp2x = np.array(bp2x)
freq_features["BandPower2X"] = bp2x

plt.figure(figsize=(10,4))
plt.plot(time_final, bp2x)
plt.title("Band Power 2X (2× shaft frequency)")
plt.xlabel("Time (s)"); plt.ylabel("Band Power")
plt.grid(True)
plt.show()
```

```python
# Build frequency-domain DataFrame
freq_df = pd.DataFrame(freq_features)
freq_df["time"] = time_final
print("Frequency domain features shape:", freq_df.shape)
freq_df.head()
```

### 3C — Time-Frequency Features (5) using Wavelet

We use a simple manual Haar-like wavelet decomposition (no pywavelets needed).

```python
# Simple manual wavelet decomposition (Haar-like)
def wavelet_decompose(signal, level=4):
    """Break signal into frequency sub-bands by repeated averaging."""
    coeffs = []
    current = signal.copy()
    for _ in range(level):
        if len(current) < 2:
            break
        n = len(current) // 2 * 2
        approx = (current[:n:2] + current[1:n:2]) / 2  
        detail = (current[:n:2] - current[1:n:2]) / 2   
        coeffs.append(detail)
        current = approx
    coeffs.append(current) 
    return coeffs

tf_features = {}
```

```python
# 1. Wavelet Energy
wavelet_energy = []
for i in range(0, len(vibration) - window_size, step):
    seg    = vibration[i:i+window_size]
    coeffs = wavelet_decompose(seg)
    energy = sum(np.sum(c**2) for c in coeffs)
    wavelet_energy.append(energy)
wavelet_energy = np.array(wavelet_energy)
tf_features["WaveletEnergy"] = wavelet_energy

plt.figure(figsize=(10,4))
plt.plot(time_final, wavelet_energy)
plt.title("Wavelet Energy")
plt.xlabel("Time (s)"); plt.ylabel("Energy")
plt.grid(True)
plt.show()
```

```python
# 2. Wavelet Entropy
wavelet_entropy = []
for i in range(0, len(vibration) - window_size, step):
    seg      = vibration[i:i+window_size]
    coeffs   = wavelet_decompose(seg)
    energies = np.array([np.sum(c**2) for c in coeffs])
    prob     = energies / (np.sum(energies) + 1e-12)
    entropy  = -np.sum(prob * np.log(prob + 1e-12))
    wavelet_entropy.append(entropy)
wavelet_entropy = np.array(wavelet_entropy)
tf_features["WaveletEntropy"] = wavelet_entropy

plt.figure(figsize=(10,4))
plt.plot(time_final, wavelet_entropy)
plt.title("Wavelet Entropy")
plt.xlabel("Time (s)"); plt.ylabel("Entropy")
plt.grid(True)
plt.show()
```

```python
# 3. Wavelet RMS
wavelet_rms = []
for i in range(0, len(vibration) - window_size, step):
    seg      = vibration[i:i+window_size]
    coeffs   = wavelet_decompose(seg)
    combined = np.concatenate(coeffs)
    wavelet_rms.append(np.sqrt(np.mean(combined**2)))
wavelet_rms = np.array(wavelet_rms)
tf_features["WaveletRMS"] = wavelet_rms

plt.figure(figsize=(10,4))
plt.plot(time_final, wavelet_rms)
plt.title("Wavelet RMS")
plt.xlabel("Time (s)"); plt.ylabel("RMS")
plt.grid(True)
plt.show()
```

```python
# 4. Wavelet Kurtosis
wavelet_kurt = []
for i in range(0, len(vibration) - window_size, step):
    seg      = vibration[i:i+window_size]
    coeffs   = wavelet_decompose(seg)
    combined = np.concatenate(coeffs)
    wavelet_kurt.append(kurtosis(combined))
wavelet_kurt = np.array(wavelet_kurt)
tf_features["WaveletKurtosis"] = wavelet_kurt

plt.figure(figsize=(10,4))
plt.plot(time_final, wavelet_kurt)
plt.title("Wavelet Kurtosis")
plt.xlabel("Time (s)"); plt.ylabel("Kurtosis")
plt.grid(True)
plt.show()
```

```python
# 5. Wavelet Variance
wavelet_var = []
for i in range(0, len(vibration) - window_size, step):
    seg      = vibration[i:i+window_size]
    coeffs   = wavelet_decompose(seg)
    combined = np.concatenate(coeffs)
    wavelet_var.append(np.var(combined))
wavelet_var = np.array(wavelet_var)
tf_features["WaveletVariance"] = wavelet_var

plt.figure(figsize=(10,4))
plt.plot(time_final, wavelet_var)
plt.title("Wavelet Variance")
plt.xlabel("Time (s)"); plt.ylabel("Variance")
plt.grid(True)
plt.show()
```

```python
# Build time-frequency DataFrame
tf_df = pd.DataFrame(tf_features)
tf_df["time"] = time_final
print("Time-Frequency features shape:", tf_df.shape)
tf_df.head()
```

## Step 4: Combine All Features into One DataFrame

```python
# Combine all three feature sets
final_df = pd.concat([
    time_df.drop(columns=["time"]),
    freq_df.drop(columns=["time"]),
    tf_df
], axis=1)

print("Final feature matrix shape:", final_df.shape)
print("Columns:", list(final_df.columns))
final_df.head()
```

## Step 5: Split into Healthy (Train) and Full (Test)

- First 40% of windows = **Healthy** (used to train PCA model)
- All windows = **Test** (used to detect fault and track degradation)

```python
n_total   = len(final_df)
n_healthy = int(0.40 * n_total)   

X_all   = final_df.drop(columns=["time"])
X_train = X_all.iloc[:n_healthy]  
X_test  = X_all              

print("Total windows  :", n_total)
print("Healthy (train):", n_healthy)
print("Test (all)     :", len(X_test))
```

## Step 6: Normalize Features

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled  = scaler.transform(X_test)
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled  = pd.DataFrame(X_test_scaled,  columns=X_test.columns)

print("Train scaled shape:", X_train_scaled.shape)
print("Test  scaled shape:", X_test_scaled.shape)
X_train_scaled.head()
```

## Step 7: PCA — Dimensionality Reduction

PCA is trained on healthy data only.
It learns the **normal pattern** of the bearing.
When the bearing degrades, data deviates from this pattern.

```python
from sklearn.decomposition import PCA
pca_full = PCA()
pca_full.fit(X_train_scaled)

plt.figure(figsize=(10,4))
plt.plot(pca_full.explained_variance_ratio_.cumsum(), marker='o', markersize=3)
plt.axhline(0.95, color='red', linestyle='--', label='95% variance')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("How many PCA components do we need?")
plt.legend()
plt.grid(True)
plt.show()
```

```python
n_components = 8

pca = PCA(n_components=n_components)
pca.fit(X_train_scaled) 

for i, var in enumerate(pca.explained_variance_ratio_):
    print(f"PC{i+1} explains {var*100:.2f}% variance")
print(f"\nTotal: {pca.explained_variance_ratio_.sum()*100:.2f}%")
```

## Step 8: PCA Score Matrix

Project all data (train + test) onto the PCA components.
This gives us the **PC scores** for each window — a new table with fewer columns.

```python
pca = PCA(n_components=8)

pca.fit(X_train_scaled)

scores_train = pca.transform(X_train_scaled)
scores_test = pca.transform(X_test_scaled)

print(scores_test.shape)
```

```python
scores_all = pca.transform(X_test_scaled)
score_cols = [f"PC{i+1}" for i in range(n_components)]
pca_df     = pd.DataFrame(scores_all, columns=score_cols)
pca_df["time"] = time_final[:len(pca_df)]

print("PCA Score Matrix shape:", pca_df.shape)
print("\nSample (first 5 rows):")
print(pca_df.head())
```

```python
# Plot each PC score over time
for col in score_cols:
    plt.figure(figsize=(10,4))
    plt.plot(pca_df["time"], pca_df[col])
    plt.title(f"{col} Score over Time")
    plt.xlabel("Time (s)")
    plt.ylabel(col)
    plt.grid(True)
    plt.show()
```

## Step 9: Feature Fusion — Global PC (GPC)

We combine all PC scores into one single value.
Each PC gets a weight based on how much variance it explains.

**GPC = w1×PC1 + w2×PC2 + w3×PC3 + ...**

This gives one trend that represents overall bearing degradation.

```python
weights = pca.explained_variance_ratio_

print("Fusion weights:")
for i, w in enumerate(weights):
    print(f"  PC{i+1}: {w:.4f}")
GPC = scores_all @ weights 

plt.figure(figsize=(10,4))
plt.plot(time_final, GPC)
plt.title("Global Principal Component (GPC) — Degradation Trend")
plt.xlabel("Time (s)")
plt.ylabel("GPC")
plt.grid(True)
plt.show()
```

## Step 10: Health Indicator (HI)

We normalize GPC to get a **Health Indicator** between 0 and 1.
- HI ≈ 0 → bearing is healthy
- HI increases → bearing is degrading
- HI → 1 → bearing is near failure

```python
# Take absolute value then normalize to [0, 1]
HI_raw = np.abs(GPC)
HI_min = HI_raw.min()
HI_max = HI_raw.max()
HI = (HI_raw - HI_min) / (HI_max - HI_min + 1e-12)
HI_smooth = pd.Series(HI).rolling(10).mean().bfill()


plt.figure(figsize=(10,5))
plt.plot(time_final, HI_smooth, color='steelblue')
plt.title("Health Indicator (HI) — 0 = Healthy, 1 = Failure")
plt.xlabel("Time (s)")
plt.ylabel("HI")
plt.grid(True)
plt.show()
```

```python
# ===== IMPROVED HI — Smooth + Monotonic =====
HI_smooth_improved = pd.Series(HI).rolling(window=5).mean().bfill()
HI_monotonic = np.maximum.accumulate(HI_smooth_improved)

plt.figure(figsize=(10, 4))
plt.plot(time_final, HI_monotonic, color='darkgreen', label='Monotonic HI')
plt.plot(time_final, HI_smooth_improved, color='steelblue', alpha=0.5, label='Smooth HI')
plt.title('Improved Health Indicator (Smooth + Monotonic)')
plt.xlabel('Time (s)')
plt.ylabel('HI')
plt.legend()
plt.grid(True)
plt.show()
```

```python

BPFO = 236.4   # Ball Pass Freq Outer Race (Hz)
BPFI = 296.9   # Ball Pass Freq Inner Race (Hz)
BSF  = 139.9   # Ball Spin Frequency (Hz)
# FTF  = 11.0    # Fundamental Train Frequency / cage (Hz)
BW   = 20      # bandwidth ± Hz around each defect frequency

def classify_fault(spectrum, freqs, threshold=0.0):
    """Rule-based fault type classifier based on defect frequency energy."""
    def band_energy(fc):
        mask = (freqs >= fc - BW) & (freqs <= fc + BW)
        return np.sum(spectrum[mask]**2) if mask.any() else 0.0

    e_bpfo = band_energy(BPFO)
    e_bpfi = band_energy(BPFI)
    e_bsf  = band_energy(BSF)
    # e_ftf  = band_energy(FTF)

    energies   = {"Outer Race": e_bpfo, "Inner Race": e_bpfi,
                  "Ball Fault": e_bsf} #"Cage Fault": e_ftf
    fault_type = max(energies, key=energies.get)
    confidence = max(energies.values()) / (sum(energies.values()) + 1e-12)
    return fault_type, confidence, energies

freqs_win = fftfreq(window_size, d=1/Fs)[:window_size//2]

fault_types   = []
confidences   = []
energy_bpfo_v = []
energy_bpfi_v = []
energy_bsf_v  = []
# energy_ftf_v  = []

for i in range(0, len(vibration) - window_size, step):
    seg      = vibration[i:i+window_size]
    spectrum = np.abs(fft(seg))[:window_size//2]
    ft, conf, energies = classify_fault(spectrum, freqs_win)
    fault_types.append(ft)
    confidences.append(conf)
    energy_bpfo_v.append(energies["Outer Race"])
    energy_bpfi_v.append(energies["Inner Race"])
    energy_bsf_v.append(energies["Ball Fault"])
    # energy_ftf_v.append(energies["Cage Fault"])

fault_types   = np.array(fault_types)
confidences   = np.array(confidences)

# Only report fault type for FAULT windows
fault_type_in_fault = fault_types[fault_mask]
from collections import Counter
type_counts = Counter(fault_type_in_fault)
print("\nFault type distribution (in fault windows):")
for ftype, cnt in type_counts.most_common():
    print(f"  {ftype:12s}: {cnt} windows ({100*cnt/n_fault:.1f}%)")

dominant_fault = type_counts.most_common(1)[0][0]
print(f"\nDominant fault type: {dominant_fault}")
```

```python
type_palette = {"Outer Race": "red", "Inner Race": "darkorange",
                "Ball Fault": "purple", "Cage Fault": "brown"}

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Defect frequency energies over time
ax = axes[0, 0]
ax.semilogy(time_final, energy_bpfo_v, color="red",        linewidth=0.8, label=f"BPFO ({BPFO}Hz)")
ax.semilogy(time_final, energy_bpfi_v, color="darkorange", linewidth=0.8, label=f"BPFI ({BPFI}Hz)")
ax.semilogy(time_final, energy_bsf_v,  color="purple",     linewidth=0.8, label=f"BSF ({BSF}Hz)")
ax.semilogy(time_final, energy_ftf_v,  color="brown",      linewidth=0.8, label=f"FTF ({FTF}Hz)")
ax.axvline(fault_onset_time, color="black", linestyle="--", linewidth=1.5, label="Fault onset")
ax.set_title("Defect Frequency Band Energies"); ax.set_xlabel("Time (s)"); ax.set_ylabel("Energy (log)")
ax.legend(fontsize=8); ax.grid(True, alpha=0.4)

# 2. Fault type colour-coded over time (fault windows only)
ax = axes[0, 1]
for ftype, col in type_palette.items():
    mask_ft = fault_mask & (fault_types == ftype)
    ax.scatter(time_final[mask_ft], HI_top[mask_ft], s=8, c=col, alpha=0.6, label=ftype)
mask_h = fault_label == 0
ax.scatter(time_final[mask_h], HI_top[mask_h], s=6, c="green", alpha=0.4, label="Healthy")
ax.axvline(fault_onset_time, color="black", linestyle="--")
ax.set_title("Fault Type Classification on HI"); ax.set_xlabel("Time (s)"); ax.set_ylabel("HI")
ax.legend(fontsize=8); ax.grid(True, alpha=0.4)

# 3. Fault type distribution bar chart
ax = axes[1, 0]
types_all   = [k for k in type_counts.most_common()]
labels_bar  = [t[0] for t in types_all]
counts_bar  = [t[1] for t in types_all]
bar_colors  = [type_palette[l] for l in labels_bar]
ax.bar(labels_bar, counts_bar, color=bar_colors)
ax.set_title("Fault Type Distribution (Fault Windows)"); ax.set_ylabel("Window Count")
ax.grid(True, axis='y', alpha=0.4)

# 4. Confidence over time
ax = axes[1, 1]
ax.scatter(time_final[fault_mask],   confidences[fault_mask],   s=6, c="red",   alpha=0.5, label="Fault windows")
ax.scatter(time_final[healthy_mask], confidences[healthy_mask], s=4, c="green", alpha=0.3, label="Healthy windows")
ax.axvline(fault_onset_time, color="black", linestyle="--")
ax.set_title("Classification Confidence"); ax.set_xlabel("Time (s)"); ax.set_ylabel("Confidence")
ax.legend(); ax.grid(True, alpha=0.4)

plt.suptitle("Fault Type Detection — NASA IMS Bearing 1", fontsize=14, fontweight='bold')
plt.tight_layout(); plt.show()
```

```python
# ── PCA on healthy region ────────────────────────────────────────────────────
n_pca_components = 6

pca_q = PCA(n_components=n_pca_components)
pca_q.fit(X_train_scaled.values)

# Project all data, then reconstruct
scores_train = pca_q.transform(X_train_scaled.values)
scores_all_q = pca_q.transform(X_test_scaled.values)

recon_all    = pca_q.inverse_transform(scores_all_q)
residuals    = X_test_scaled.values - recon_all
Q_stat       = np.sum(residuals**2, axis=1)   # SPE = Q statistic

recon_train  = pca_q.inverse_transform(scores_train)
resid_train  = X_train_scaled.values - recon_train
Q_train      = np.sum(resid_train**2, axis=1)

# ── UCL threshold (99% confidence on healthy Q) ──────────────────────────────
Q_threshold = np.percentile(Q_train, 99)

# ── Smooth Q for cleaner visualisation ───────────────────────────────────────
Q_smooth = pd.Series(Q_stat).rolling(5).mean().bfill().values

print(f"Q threshold (99th pct on healthy): {Q_threshold:.6f}")
print(f"Max Q (full run): {Q_stat.max():.4f}")
```

```python
# ── UNSUPERVISED PLOT — matches the figure style ─────────────────────────────

import matplotlib.patches as mpatches

fig, ax = plt.subplots(figsize=(12, 5))

# NOR shaded region (below threshold)
ax.axhspan(Q_stat.min()*0.5, Q_threshold, alpha=0.15, color="green", label="NOR")

# Training data (healthy portion) — blue
Q_train_plot = Q_smooth[:n_healthy]
t_train_plot = time_final[:n_healthy]
ax.semilogy(t_train_plot, Q_train_plot, color="blue", linewidth=0.9, label="Training data")

# Testing data — up to fault onset: blue; from fault onset onwards: ALL red
t_test_plot = time_final[n_healthy:]
Q_test_plot = Q_smooth[n_healthy:]

# Split at fault onset
pre_fault_mask_test  = time_final[n_healthy:] < fault_onset_time
post_fault_mask_test = ~pre_fault_mask_test   # everything from fault onset = RED

if pre_fault_mask_test.any():
    ax.semilogy(time_final[n_healthy:][pre_fault_mask_test],
                Q_smooth[n_healthy:][pre_fault_mask_test],
                color="blue", linewidth=0.9)

if post_fault_mask_test.any():
    ax.semilogy(time_final[n_healthy:][post_fault_mask_test],
                Q_smooth[n_healthy:][post_fault_mask_test],
                color="red", linewidth=0.9, label="Testing data (fault)")

# Q threshold line
ax.axhline(Q_threshold, color="black", linestyle="--", linewidth=1.5, label=f"-Q threshold ({Q_threshold:.4f})")

# Fault onset vertical marker
ax.axvline(fault_onset_time, color="darkorange", linestyle=":", linewidth=1.5, label=f"Fault onset ({fault_onset_time:.0f}s)")
ax.set_title("Unsupervised Trend Analysis — PCA Q-Statistic (SPE)", fontsize=13, fontweight='bold')
ax.set_xlabel("Time (Seconds)", fontsize=11)
ax.set_ylabel("Q", fontsize=11)

patches = [
    mpatches.Patch(color="blue",  label="Training data"),
    mpatches.Patch(color="red",   label="Testing data"),
    plt.Line2D([0],[0], color="black", linestyle="--", label="-Q threshold"),
    mpatches.Patch(color="green", alpha=0.3, label="NOR"),
]
ax.legend(handles=patches, loc="upper left", fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("unsupervised_Q_chart.png", dpi=150, bbox_inches='tight')
plt.show()
print("Saved: unsupervised_Q_chart.png")
```

## Kernel PCA — Alternative Dimensionality Reduction

Kernel PCA with RBF kernel captures non-linear structure in the data,
compared to linear PCA. We compare HI quality and RUL prediction performance.

```python
# ===== KERNEL PCA =====
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=6, kernel='rbf', gamma=0.1)
X_kpca_train = kpca.fit_transform(X_train_scaled)
X_kpca_test  = kpca.transform(X_test_scaled)

print('Shape after KPCA (train):', X_kpca_train.shape)
print('Shape after KPCA (test) :', X_kpca_test.shape)
```

```python
# ===== HEALTH INDICATOR FROM KPCA =====
HI_kpca_raw = np.sum(np.abs(X_kpca_test), axis=1)
HI_kpca_raw = (HI_kpca_raw - HI_kpca_raw.min()) / (HI_kpca_raw.max() - HI_kpca_raw.min() + 1e-12)
HI_kpca_smooth = pd.Series(HI_kpca_raw).rolling(window=5).mean().bfill()
HI_kpca_monotonic = np.maximum.accumulate(HI_kpca_smooth)

plt.figure(figsize=(10, 4))
plt.plot(time_final, HI_kpca_monotonic, color='purple', label='KPCA HI (monotonic)')
plt.plot(time_final, HI_smooth, color='steelblue', linestyle='--', alpha=0.6, label='PCA HI')
plt.title('Health Indicator: PCA vs Kernel PCA')
plt.xlabel('Time (s)')
plt.ylabel('HI')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# ===== RANDOM FOREST USING KPCA FEATURES =====
from sklearn.model_selection import train_test_split as tts
from sklearn.metrics import r2_score


X_train_kpca_reg, X_test_kpca_reg, y_train_kpca_reg, y_test_kpca_reg = tts(
    X_kpca_test, RUL_true, test_size=0.2, random_state=42
)

rf_kpca = RandomForestRegressor(n_estimators=150, max_depth=12, random_state=42)
rf_kpca.fit(X_train_kpca_reg, y_train_kpca_reg)
y_pred_kpca = rf_kpca.predict(X_test_kpca_reg)

mae_kpca  = mean_absolute_error(y_test_kpca_reg, y_pred_kpca)
rmse_kpca = np.sqrt(mean_squared_error(y_test_kpca_reg, y_pred_kpca))
r2_kpca   = r2_score(y_test_kpca_reg, y_pred_kpca)

print('KPCA + RF MODEL')
print(f'  MAE  : {mae_kpca:.2f} seconds')
print(f'  RMSE : {rmse_kpca:.2f} seconds')
print(f'  R2   : {r2_kpca:.4f}')
```
