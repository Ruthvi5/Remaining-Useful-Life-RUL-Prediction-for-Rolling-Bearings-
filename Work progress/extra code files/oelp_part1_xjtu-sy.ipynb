```python
import os
import pandas as pd
import numpy as np

folder = "XJTU-SY_Bearing_Datasets 2/37.5Hz11kN/Bearing2_1"

all_data = []

for file in sorted(os.listdir(folder)):  
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(folder, file), header=None)

        df.columns = df.iloc[0]
        df = df.drop(0).reset_index(drop=True)

        df = df.astype(float)

        all_data.append(df)

combined_df = pd.concat(all_data, ignore_index=True)
combined_df.shape

```

```python
fs = 37.5  # Hz
time = np.arange(len(combined_df)) / fs

```

```python
N = 3000

plt.figure(figsize=(10, 4))
plt.plot(time[:N], combined_df["Horizontal_vibration_signals"][:N])
plt.xlabel("Time (seconds)")
plt.ylabel("Acceleration")
plt.title("Zoomed Combined Signal (First 3000 Samples)")
plt.grid(True)
plt.show()

```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

folder = "XJTU-SY_Bearing_Datasets 2/37.5Hz11kN/Bearing2_1"
fs = 37.5  # Hz

plt.figure(figsize=(12, 5))

current_time = 0

for file in sorted(os.listdir(folder)):
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(folder, file), header=None)

        df.columns = df.iloc[0]
        df = df.drop(0).reset_index(drop=True)
        df = df.astype(float)

        t = np.arange(len(df)) / fs + current_time

        plt.plot(t, df["Vertical_vibration_signals"])

        current_time = t[-1]

plt.xlim(3400, 3600)

plt.xlabel("Time (seconds)")
plt.ylabel("Acceleration")
plt.title("Vertical Vibration (Zoomed: 3400–3600 sec)")
plt.grid(True)
plt.show()
```
