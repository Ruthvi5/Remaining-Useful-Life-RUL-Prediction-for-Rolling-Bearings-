```python
import os
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
data_dir = Path(
    "/Users/maramruthvi/.cache/kagglehub/datasets/"
    "vinayak123tyagi/bearing-dataset/versions/1/"
    "2nd_test/2nd_test"
)
```

```python
data_paths = sorted(data_dir.glob("*"))

print("Total number of files:", len(data_paths))
print("\nFirst 5 files:")
for p in data_paths[:5]:
    print(p.name)
```

```python
merged_data = []
timestamps = []

for file_path in data_paths:
    filename = file_path.name
    
    # Read vibration file (tab separated)
    dataset = pd.read_csv(file_path, sep="\t", header=None)
    
    # Mean Absolute Value for each bearing
    mav = dataset.abs().mean().values  # shape (4,)
    
    merged_data.append(mav)
    timestamps.append(pd.to_datetime(filename, format="%Y.%m.%d.%H.%M.%S"))
```

```python
merged_df = pd.DataFrame(
    merged_data,
    columns=["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"]
)

merged_df.index = timestamps
merged_df = merged_df.sort_index()
```

```python
merged_df.to_csv("merged_Dataset_BearingTest_2.csv")
merged_df.head()
```

```python
import matplotlib.pyplot as plt

for bearing in merged_df.columns:
    plt.figure(figsize=(10,4))
    plt.plot(merged_df.index, merged_df[bearing])
    plt.xlabel("Time")
    plt.ylabel("Mean Absolute Acceleration")
    plt.title(f"{bearing} – Acceleration vs Time")
    plt.grid(True)
    plt.show()
```
